```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
import cohere as co

client = co.Client(os.getenv("COHERE_API_KEY"))
```

```python
prompt = '''Whenever I see a cat, :
complete the sentence above under 20 words'''
```

```python
response = client.chat(
    model="command-a-03-2025",
    message=prompt,
    temperature=1.6,
    k=20,
    p=0.75
)

print(response.text)
```
